```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/Image_compression_results/kodak_metrics.csv")


# ---- Average metrics per agent ----
grouped = df.groupby("Agent").mean(numeric_only=True)

# -------- PSNR --------
plt.figure()
grouped["PSNR"].plot(kind="bar", title="Average PSNR Comparison")
plt.ylabel("PSNR (dB)")
plt.tight_layout()
plt.show()
```

```python
# -------- RD: PSNR vs LPIPS --------
plt.figure(figsize=(7,5))

for agent in df["Agent"].unique():
    sub = df[df["Agent"] == agent]
    plt.scatter(sub["PSNR"], sub["LPIPS"], label=agent, s=40)

plt.xlabel("PSNR (dB)")
plt.ylabel("LPIPS (Lower is Better)")
plt.title("Pixel Fidelity vs Perceptual Quality (PSNR vs LPIPS)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# -------- RD: LPIPS vs BPP --------
plt.figure(figsize=(7,5))

for agent in df["Agent"].unique():
    sub = df[df["Agent"] == agent]
    plt.scatter(sub["BPP"], sub["LPIPS"], label=agent, s=40)

plt.xlabel("Bits Per Pixel (BPP)")
plt.ylabel("LPIPS (Lower is Better)")
plt.title("Perceptual Rate–Distortion (LPIPS vs BPP)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# -------- RD: PSNR vs BPP --------
plt.figure()
for agent in df["Agent"].unique():
    subset = df[df["Agent"] == agent]
    plt.scatter(subset["BPP"], subset["PSNR"], label=agent)

plt.xlabel("Bits Per Pixel (BPP)")
plt.ylabel("PSNR (dB)")
plt.title("Rate–Distortion Curve (PSNR vs BPP)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# -------- BPP --------
plt.figure()
grouped["BPP"].plot(kind="bar", title="Average Bits Per Pixel (BPP)")
plt.ylabel("BPP")
plt.tight_layout()
plt.show()
```

```python

# -------- LPIPS --------
plt.figure()
grouped["LPIPS"].plot(kind="bar", title="Average LPIPS (Lower is Better)")
plt.ylabel("LPIPS")
plt.tight_layout()
plt.show()
```

```python
# -------- SSIM --------
plt.figure()
grouped["SSIM"].plot(kind="bar", title="Average SSIM Comparison")
plt.ylabel("SSIM")
plt.tight_layout()
plt.show()
```
